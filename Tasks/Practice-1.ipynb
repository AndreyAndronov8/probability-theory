```python
import scipy.special as sc
```

### Пример 1.8.
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены:

>а) различные призы;

$\tilde{A}_{10}^5 = 10^5 = 100000$

```python
display(10**5)
```

>б) одинаковые призы?

$\tilde{C}_{10}^5 = C_{10+5-1}^5 = C_{14}^5 = \frac{14 \cdot 13 \cdot 12 \cdot 11 \cdot 10}{1 \cdot 2 \cdot 3 \cdot 4 \cdot 5} = 2002$


```python
display(round(sc.comb(10+5-1, 5)))
```

### Пример 1.9.
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

$P_7(3; 2; 2) = \frac{7!}{3! \, 2! \, 2!} = 210$


```python
display(round(sc.factorial(7)/
              (sc.factorial(3)*
               sc.factorial(2)*
               sc.factorial(2))))
```

## Тема 1.6. Непосредственное вычисление вероятностей.
---

### Пример 1.10.
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой:

>а) 3 карточки; Какова вероятность того, что получится слово: «ТОР»;

$P(A) = \frac{m}{n} = \frac{1}{A_6^3} = \frac{1}{6 \cdot 5 \cdot 4} = \frac{1}{120}$

```python
display(1/120, 1/sc.perm(6, 3))
```

>б) все 6 карточек. Какова вероятность того, что получится слово: «ТЕОРИЯ»?

$P(B) = \frac{m}{n} = \frac{1}{P_6} = \frac{1}{6!} = \frac{1}{720}$

```python
display(1/720, 1/sc.factorial(6))
```

### Пример 1.11.
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

$P(B) = \frac{m}{n} = \frac{P_3 \cdot P_2}{P_6} = \frac{3! \, 2!}{6!} = \frac{1}{60}$


```python
display(1/60, sc.factorial(3)*sc.factorial(2)/sc.factorial(6))
```

### Пример 1.16.
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что:

> а) никакие два лица одного пола не сядут рядом;

$P(A) = \frac{m_1}{n} = \frac{2(5!)^2}{10!} = \frac{2 \cdot 120^2}{3\,628\,800} = 0,00794$


```python
display(round(2*sc.factorial(5)**2/sc.factorial(10), 5))
```

> б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

$P(B) = \frac{m_2}{n} = \frac{5! \, 2^5}{10!} = \frac{120 \cdot 32}{3\,628\,800} = 0,00106$

```python
display(round(sc.factorial(5)*2**5/sc.factorial(10), 5))
```

### Пример 1.17.
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали:

> а) в два купе;

$P(A) = \frac{m_1}{n} = \frac{C_9^2 \cdot C_8^7}{C_{36}^7} = \frac{36 \cdot 8}{8\,347\,680} = 0,0000345.$

```python
display(sc.comb(9, 2)*sc.comb(8, 7)/sc.comb(36, 7))
```

> б) в семь купе;

$P(B) = \frac{m_2}{n} = \frac{C_9^7 \cdot 4^7}{C_{36}^7} = \frac{36 \cdot 16\,384}{8\,347\,680} = 0,07066.$

```python
display(sc.comb(9, 7)*4**7/sc.comb(36, 7))
```

> в) в три купе.

$\begin{aligned}m_3 &= C_9^3 (6 C_4^4 C_4^2 C_4^1 + 3 C_4^3 C_4^3 C_4^1 + 3 C_4^3 C_4^2 C_4^2) = \\&= 84 (6 \cdot 24 + 3 \cdot 64 + 3 \cdot 144) = 84 \cdot 768 = 64\,512.\end{aligned}$

$P(D) = \frac{m_3}{n} = \frac{64\,512}{8\,347\,680} = 0,00773.$

```python
m3 = round(sc.comb(9, 3)*(6*sc.comb(4, 4)*sc.comb(4, 2)*sc.comb(4, 1) + 3*sc.comb(4, 3)*sc.comb(4, 3)*sc.comb(4, 1) + 3*sc.comb(4, 3)*sc.comb(4, 2)*sc.comb(4, 2)))
display(m3, m3/sc.comb(36, 7))
```
