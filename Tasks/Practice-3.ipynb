# Моделирование задачи с шарами

> Решить классически и смоделировать решение задачи. В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

```python
import numpy as np
import pandas as pd
import scipy.special as sc
```

```python
urn1 = {"white": 15, "black": 5}
urn2 = {"white": 6, "black": 1}

shift = 3

take_white = 1
take_black = 0

P = 0
for i in range(0, shift + 1):
    P += (sc.comb(urn1["white"], i)) \
        * (sc.comb(urn1["black"], shift - i)) \
        / (sc.comb(urn1["white"] + urn1["black"], shift)) \
        * (sc.comb(urn2["white"] + i, take_white)) \
        * (sc.comb(urn2["black"] + shift - i, take_black)) \
        / (sc.comb(urn2["white"] + urn2["black"] + shift, take_white + take_black))

print(f"Теоретическая вероятность равна: {P:0.4f}")
```

```python
n = 1_000_000

urna1 = "w" * urn1["white"] + "b" * urn1["black"]
urna2 = "w" * urn2["white"] + "b" * urn2["black"]
print(urna1)
print(urna2)
```

```python
def shift_ball(urna1, urna2):
    i = np.random.randint(0, len(urna1))
    urna2 += urna1[i]
    urna1 = urna1[:i] + urna1[i+1:]
    return urna1, urna2

result = []

for i in range(n):
    urna1 = "w" * urn1["white"] + "b" * urn1["black"]
    urna2 = "w" * urn2["white"] + "b" * urn2["black"]
    for j in range(shift):
        urna1, urna2 = shift_ball(urna1, urna2)
    
    taken = []
    for j in range(take_white + take_black):
        ind = np.random.randint(0, len(urna2))
        taken.append(urna2[ind])
        urna2 = urna2[:ind] + urna2[ind+1:]
    
    result.append(taken)
```

```python
df = pd.DataFrame(result)
df.sample(10)
```

```python
good_df = df[(df == 'w').sum(axis=1) == take_white]
print(f"Теоретическая вероятность равна: {P:0.5f}")
print(f"Экспериментальная вероятность: {len(good_df)/len(df):0.5f}")
```
