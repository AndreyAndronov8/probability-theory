### Пример 2.9.
> В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того что из 9 пакетов акций в результате торгов по первоначально заявленной цене: 
> 1) не будут проданы 5 пакетов; 
> 2) будет продано: \
    а) менее 2 пакетов; \
    б) не более 2 пакетов; \
    в) хотя бы 2 пакета; \
    г) наивероятнейшее число пакетов.

```python
# 1) не будут проданы 5 пакетов

m = 5
n = 9
p = 0.8
q = 1 - p

print(f"P = {sc.binom.pmf(m, n, p):.4f} (Бернулли)" )
print(f"np = {n * p:.4f} <= 10, но p > 0.1, нельзя  использовать формулу Пуассона" )
print(f"npq = {n * p * q:.4f} < 20, нельзя использовать теорему Муавра-Лапласа" )
```

```python
# 2) а) будет продано менее 2 пакетов

n = 9
p = 0.2
q = 1 - p

print(f"P = {sc.binom.pmf(0, n, p) + sc.binom.pmf(1, n, p):.4f} (Бернулли)")
print(f"np = {n * p:.4f} <= 10, но p > 0.1, нельзя  использовать формулу Пуассона")
print(f"npq = {n * p * q:.4f} < 20, нельзя использовать теорему Муавра-Лапласа")
```

```python
# 2) б) будет продано не более 2 пакетов

n = 9
p = 0.2
q = 1 - p

print(f"P = {sc.binom.pmf(0, n, p) + sc.binom.pmf(1, n, p) + sc.binom.pmf(2, n, p):.4f} (Бернулли)" )
print(f"np = {n * p:.4f} <= 10, но p > 0.1, нельзя  использовать формулу Пуассона")
print(f"npq = {n * p * q:.4f} < 20, нельзя использовать теорему Муавра-Лапласа")
```

```python
# 2) в) будет продано хотя бы 2 пакета

n = 9
p = 0.2
q = 1 - p

print(f"P = {1 - (sc.binom.pmf(0, n, p) + sc.binom.pmf(1, n, p)):.4f} (Бернулли)")
print(f"np = {n * p:.4f} <= 10, но p > 0.1, нельзя  использовать формулу Пуассона")
print(f"npq = {n * p * q:.4f} < 20, нельзя использовать теорему Муавра-Лапласа")
```

```python
# 2) г) наивероятнейшее число событий

n = 9
p = 0.2
q = 1 - p

print(f"{n * p - q} <= m0 <= {n * p + p}")

print(f"P(1) = {sc.binom.pmf(1, n, p):.4f} (Бернулли)")
print(f"P(2) = {sc.binom.pmf(2, n, p):.4f} (Бернулли)")
print(f"np = {n * p:.4f} <= 10, но p > 0.1, нельзя  использовать формулу Пуассона")
print(f"npq = {n * p * q:.4f} < 20, нельзя использовать теорему Муавра-Лапласа")
```

### Пример 2.10.
> Завод отправил на базу 10 000 стандартных изделий. Среднее число изделий, повреждаемых при транспортировке,составляет 0,02%. Найти вероятность того, что из 10 000 изделий:
> 1) будет повреждено: \
>   а) 3; \
>   б) по крайней мере 3; \
> 2) не будет повреждено: \
>   а) 9997; \
>   б) хотя бы 9997.

```python
# 1) а) 3

m = 3
n = 10000
p = 0.0002
q = 1 - p

print(f"P = {sc.binom.pmf(m, n, p):.4f} (Бернулли)" )
print(f"np = {n * p:.4f} <= 10, и p < 0.1, можно использовать формулу Пуассона" )
print(f"P = {sc.poisson.pmf(k=m, mu=n*p):.4f} (Пуассон)" )
print(f"npq = {n * p * q:.4f} < 20, нельзя использовать теорему Муавра-Лапласа" )
```

```python
# 1) б) по крайней мере 3

n = 10000
p = 0.0002
q = 1 - p

print(f"P = {1 - (sc.binom.pmf(0, n, p) + sc.binom.pmf(1, n, p) + sc.binom.pmf(2, n, p)):.4f} (Бернулли)" )
print(f"np = {n * p:.4f} <= 10, и p < 0.1, можно использовать формулу Пуассона" )
print(f"P = {1 - (sc.poisson.pmf(0, n*p) + sc.poisson.pmf(1, n*p) + sc.poisson.pmf(2, n*p)):.4f} (Пуассон)" )
print(f"npq = {n * p * q:.4f} < 20, нельзя использовать теорему Муавра-Лапласа" )
```

```python
# 2) a) 9997

m = 9997
n = 10000
q = 0.0002
p = 1 - q

print(f"P = {sc.binom.pmf(m, n, p):.4f} (Бернулли)" )
print(f"Равносильно P = {sc.binom.pmf(n - m, n, q):.4f} (Бернулли)" )
print(f"np = {n * p:.4f} > 10, нельзя использовать формулу Пуассона" )
print(f"npq = {n * p * q:.4f} < 20, нельзя использовать теорему Муавра-Лапласа" )
```

```python
# 2) б) хотя бы 9997

n = 10000
p = 0.0002
q = 1 - p

print(f"P = {sc.binom.pmf(0, n, p) + sc.binom.pmf(1, n, p) + sc.binom.pmf(2, n, p) + sc.binom.pmf(3, n, p):.4f} (Бернулли)" )
print(f"np = {n * p:.4f} <= 10, p < 0.1, можно использовать формулу Пуассона" )
print(f"P = {sc.poisson.pmf(0, n * p) + sc.poisson.pmf(1, n * p) + sc.poisson.pmf(2, n * p) + sc.poisson.pmf(3, n * p):.4f} (Пуассона)" )
print(f"npq = {n * p * q:.4f} < 20, нельзя использовать теорему Муавра-Лапласа" )
```

### Пример 2.11.
> По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины: \
> а) 480 предприятий; \
> б) наивероятнейшее число предприятий; \
> в) не менее 480; \
> г) от 480 до 520.

```python
# а) 480 предприятий

m = 480
n = 1000
p = 0.5
q = 1 - p

x = (m - n * p) / np.sqrt(n * p * q)

print(f"P = {sc.binom.pmf(m, n, p):.4f} (Бернулли)" )
print(f"np = {n * p:.4f} > 10, нельзя использовать формулу Пуассона" )
print(f"npq = {n * p * q:.4f} >= 20, можно использовать теорему Муавра-Лапласа" )
print(f"P = {sc.norm.pdf(x) / np.sqrt(n * p * q):.4f} (Муавр-Лаплас)")
```

```python
# б) наивероятнейшее число предприятий

print(f"{n * p - q} <= m0 <= {n * p + p}")
m = 500

n = 1000
p = 0.5
q = 1 - p

x = (m - n * p) / np.sqrt(n * p * q)

print(f"P = {sc.binom.pmf(m, n, p):.4f} (Бернулли)" )
print(f"np = {n * p:.4f} > 10, нельзя использовать формулу Пуассона" )
print(f"npq = {n * p * q:.4f} >= 20,  можно использовать теорему Муавра-Лапласа" )
print(f"P = {sc.norm.pdf(x) / np.sqrt(n * p * q):.4f} (Муавр-Лаплас)")
```

```python
# в) не менее 480

n = 1000
p = 0.5
q = 1 - p

a = 480
b = n

x1 = (a - n * p) / np.sqrt(n * p * q)
x2 = (b - n * p) / np.sqrt(n * p * q)

print(f"P = {sum(sc.binom.pmf(m, n, p) for m in range(a, b+1)):.4f} (Бернулли)" )
print(f"np = {n * p:.4f} > 10, нельзя использовать формулу Пуассона" )
print(f"npq = {n * p * q:.4f} >= 20,  можно использовать теорему Муавра-Лапласа" )
print(f"P = {sc.norm.cdf(x2) - sc.norm.cdf(x1):.4f} (Муавр-Лаплас)")
```

```python
# г) от 480 до 520

n = 1000
p = 0.5
q = 1 - p

a = 480
b = 520

x1 = (a - n * p) / np.sqrt(n * p * q)
x2 = (b - n * p) / np.sqrt(n * p * q)

print(f"P = {sum(sc.binom.pmf(m, n, p) for m in range(a, b+1)):.4f} (Бернулли)" )
print(f"np = {n * p:.4f} > 10, нельзя использовать формулу Пуассона" )
print(f"npq = {n * p * q:.4f} >= 20,  можно использовать теорему Муавра-Лапласа" )
print(f"P = {sc.norm.cdf(x2) - sc.norm.cdf(x1):.4f} (Муавр-Лаплас)")
```
