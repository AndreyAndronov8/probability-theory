## Комбинаторика

### Перестановки

```python
# Входные данные
n = 1
nk = [] # Для перестановок с повторениями 
```

#### Без повторений или определённым количеством повторений

```python
import scipy.special as sc

ans = sc.factorial(n)
for k in range(len(nk)):
    ans /= sc.factorial(nk[k])
print(ans)
```

#### С любым количеством повторений

```python
print(n**n)
```

### Моделирование перестановок

```python
# Входные данные
items = ''
```

#### Без повторений

```python
import itertools

n = len(items)

sequences = list(itertools.permutations(items, r=n))
length = len(sequences)

counter = 0

for i in range(length):
    if True:
        counter += 1

print(counter)
```

#### С повторениями

```python
import itertools

n = len(items)

sequences = list(itertools.product(items, repeat=n))
length = len(sequences)

counter = 0

for i in range(length):
    if True:
        counter += 1

print(counter)
```

### Моделирование размещений

```python
import itertools

sequences = list(itertools.permutations(items, r=k))
length = len(sequences)

counter = 0

for i in range(length):
    if True:
        counter += 1

print(counter)
```

#### С повторениями

```python
import itertools

if k <= 10:
    sequences = list(itertools.product(items, repeat=k))
    length = len(sequences)

    counter = 0

    for i in range(length):
        if True:
            counter += 1

    print(counter)
```

### Моделирование сочетаний

```python
# Входные данные
items = 'A'
k = 1
```

#### Без повторений

```python
import itertools

sequences = list(itertools.combinations(items, r=k))
length = len(sequences)

counter = 0

for i in range(length):
    if True:
        counter += 1

print(counter)
```

#### С повторениями

```python
import itertools

sequences = list(itertools.combinations_with_replacement(items, r=k))
length = len(sequences)

counter = 0

for i in range(length):
    if True:
        counter += 1

print(counter)
```

### Теорема Муавра - Лапласа (локальная)

```python
# Входные данные
n = 400
m = 80
p = 0.2
```

```python
import scipy.stats as sc
q = 1 - p
print(f"Использовать формулу Пуассона {"можно" if n*p*q >= 20 else "нельзя"}")
x = (m - n * p) / (n * p * q) ** (1/2)
print(sc.norm.pdf(x) / (n * p * q) ** (1/2))
```

### Теорема Муавра - Лапласа (интегральная)

```python
# Входные данные
n = 400
p = 0.2

# Границы интервала
a = 70
b = 100
```

```python
from scipy.stats import binom, norm

mu = n * p
sigma = (n * p * q) ** (1/2)

normal_dist = norm(loc=mu, scale=sigma)

# # Точный расчет через биномиальное распределение
binomial_dist = binom(n, p)
prob_binomial = binomial_dist.cdf(b) - binomial_dist.cdf(a - 1)
print(f"Точная вероятность (биномиальная): {prob_binomial}")

# Вероятность в интервале [a, b] для непрерывной величины.
prob_normal_approx = normal_dist.cdf(b) - normal_dist.cdf(a)
print(f"Вероятность (нормальная аппроксимация без поправки): {prob_normal_approx}")

# Используем поправку на непрерывность (поправка Йейтса): для P(a<=X<=b) берем [a-0.5, b+0.5]
prob_normal_approx = normal_dist.cdf(b + 0.5) - normal_dist.cdf(a - 0.5)
print(f"Вероятность (нормальная аппроксимация с поправкой): {prob_normal_approx}")
```
